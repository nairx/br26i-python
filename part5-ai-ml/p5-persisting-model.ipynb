```python
from sklearn.linear_model import LinearRegression
import joblib
# Training data
X = [[1], [2], [3], [4], [5]]
y = [10, 20, 30, 40, 50]
model = LinearRegression()
model.fit(X,y)
joblib.dump(model,"my_model.pkl")
print("Model trained and saved!")

```

```python
import joblib

model = joblib.load("my_model.pkl")

prediction = model.predict([[6]])

print(prediction)

```
