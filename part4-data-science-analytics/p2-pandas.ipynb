```python
import matplotlib.pyplot as plt
x = ["2021","2022","2023","2024","2025"]
y = [10,20,25,30,40]
plt.plot(x,y,color="green",linestyle="--",marker="o",label="Toyota")
plt.title("Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = ["2021","2022","2023","2024","2025"]
y = [10,20,25,30,40]
z = [12,25,35,20,10]
plt.plot(x,y,color="green",linestyle="--",marker="o",label="Toyota")
plt.plot(x,z,color="blue",linestyle="--",marker="o",label="Kia")
plt.title("Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = ["2021","2022","2023","2024","2025"]
y = [10,20,25,30,40]
plt.bar(x,y,color="green")
plt.title("Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.savefig("sales.png")
plt.show()
```
